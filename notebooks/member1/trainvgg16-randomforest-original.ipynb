```python
import os
import glob
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
feature_dir = "/kaggle/input/datasets/huytrnnguynlm/original-feature"

X_train_rf = np.load(os.path.join(feature_dir, "X_train_rf.npy"))
y_train_rf = np.load(os.path.join(feature_dir, "y_train_rf.npy"))

X_test_rf = np.load(os.path.join(feature_dir, "X_test_rf.npy"))
y_test_rf = np.load(os.path.join(feature_dir, "y_test_rf.npy"))

print("X_train_rf:", X_train_rf.shape)
print("y_train_rf:", y_train_rf.shape)
print("X_test_rf:", X_test_rf.shape)
print("y_test_rf:", y_test_rf.shape)
```

```python
param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [10, 20],
    "min_samples_split": [2, 10],
    "min_samples_leaf": [1, 5],
    "max_features": ["sqrt"],
    "bootstrap": [True]
}

rf = RandomForestClassifier(
    n_jobs=1,
    random_state=42
)

grid = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=3,
    scoring="accuracy",
    verbose=3,
    n_jobs=1
)

grid.fit(X_train_rf, y_train_rf)

print("\nBest params:", grid.best_params_)
print(f"Best CV accuracy: {grid.best_score_:.4f}")
```

```python
best_rf = grid.best_estimator_

# Dự đoán trên train và test
train_pred = best_rf.predict(X_train_rf)
test_pred = best_rf.predict(X_test_rf)

# Accuracy
print("Train Accuracy:", accuracy_score(y_train_rf, train_pred))
print("Test Accuracy:", accuracy_score(y_test_rf, test_pred))

print("\nClassification Report trên tập Train:")
print(classification_report(y_train_rf, train_pred))

print("\nConfusion Matrix trên tập Train:")
print(confusion_matrix(y_train_rf, train_pred))

print("\nClassification Report trên tập Test:")
print(classification_report(y_test_rf, test_pred))

print("\nConfusion Matrix trên tập Test:")
print(confusion_matrix(y_test_rf, test_pred))
```
